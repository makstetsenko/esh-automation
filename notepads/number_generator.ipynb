```python
import random

def random_numbers(length: int, average: int, spread: int = 10) -> list[int]:
    numbers = []

    for _ in range(length // 2):
        deviation = random.randint(0, spread)
        numbers.extend([average - deviation, average + deviation])

    if length % 2:
        numbers.append(average)

    random.shuffle(numbers)
    return numbers
```

```python
random_numbers(length=10, average=9, spread=2)
```
